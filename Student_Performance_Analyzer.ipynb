```python
from sklearn.tree import DecisionTreeClassifier

# AI training data
X = [
    [95, 90, 92],
    [88, 85, 90],
    [80, 78, 82],
    [75, 70, 72],
    [65, 60, 68],
    [55, 50, 58],
    [45, 40, 50],
    [92, 95, 94],
    [70, 75, 73],
    [60, 65, 62]
]

y = [
    "Excellent",
    "Excellent",
    "Very Good",
    "Good",
    "Pass",
    "Needs Improvement",
    "Needsj Improvement",
    "Excellent",
    "Good",
    "Pass"
]

# Train AI model
model = DecisionTreeClassifier()
model.fit(X, y)

# Student data
name = "Haya"
math = 90
english = 75
science = 85

# Calculate average
average = (math + english + science) / 3

# AI prediction
prediction = model.predict([[math, english, science]])[0]

# Subjects
grades = {
    "Math": math,
    "English": english,
    "Science": science
}

strongest_subject = max(grades, key=grades.get)
weakest_subject = min(grades, key=grades.get)

# Status
status = "Passed" if average >= 60 else "Failed"

# AI recommendation
if prediction == "Excellent":
    recommendation = "Excellent performance! Keep challenging yourself."
elif prediction == "Very Good":
    recommendation = f"Very good! Focus more on {weakest_subject}."
elif prediction == "Good":
    recommendation = f"Good performance. Practice more in {weakest_subject}."
elif prediction == "Pass":
    recommendation = f"You passed. Focus on improving {weakest_subject}."
else:
    recommendation = f"Needs improvement. Practice {weakest_subject} regularly."

# Report
print("=================================")
print("   STUDENT PERFORMANCE ANALYZER")
print("=================================")

print("Student:", name)
print("Math:", math)
print("English:", english)
print("Science:", science)
print("Average:", round(average, 2))
print("AI Prediction:", prediction)
print("Status:", status)
print("Strongest Subject:", strongest_subject)
print("Weakest Subject:", weakest_subject)

print("\nAI Recommendation:")
print(recommendation)

print("\n=================================")
print("Analysis Completed!")
print("=================================")
```
